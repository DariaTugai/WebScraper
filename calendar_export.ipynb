# Packages import

```python
import os
import icalendar
import pandas as pd
from icalendar import Calendar, Event, vDatetime
from uuid import uuid4
import datetime
import uuid
```

# Calendar export

```python
group_name = input("Enter group name: ")
```

```python
classes = pd.read_csv(f"schedules/{group_name}.csv", encoding="UTF-8")
```

```python
calendar = Calendar()
calendar["dtstart"] = "TZID=Europe/Warsaw:2026233T00000"
calendar["dtend"] = "TZID=Europe/Warsaw:20260930T235999"
calendar["summary"] = f"Schedule for {group_name} group"
calendar["PRODID"] = "-//UEK//Apollo to iCalendar export//EN"
calendar["Version"] = "2.0"
```

```python
for index, row in classes.iterrows():
    event = Event()
    event["dtstart"] = row["Termin"].replace("-", "")+"T"+row["Start time"].replace(":", "")+"00"
    event["dtend"] = row["Termin"].replace("-", "")+"T"+row["End time"].replace(":", "")+"00"
    event["summary"] = row["Przedmiot"]
    event["location"] = row["Sala"]
    event["description"] = row["Nauczyciel"]
    event["UID"] = uuid.uuid4()
    event["dtstamp"] = vDatetime(datetime.datetime.now())
    calendar.add_component(event)
```

```python
if not os.path.exists("./calendars"):
    os.mkdir("./calendars")
```

```python
with open(f"calendars/{group_name}.ics", "wb") as cf:
    cf.write(calendar.to_ical())
```
